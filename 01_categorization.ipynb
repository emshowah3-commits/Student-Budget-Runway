```python
from datasets import load_dataset
from sklearn.model_selection import train_test_split

ds = load_dataset("DoDataThings/us-bank-transaction-categories")
print(ds)                                  # shows one split: train, 16,000 rows

df = ds["train"].to_pandas()
print(df.shape)                            # (16000, 2)
print(df["category"].value_counts())       # about 1,000 per category
print(df["description"].duplicated().sum())  # exact duplicates

df = df.drop_duplicates(subset="description").reset_index(drop=True)

train_df, test_df = train_test_split(
    df, test_size=0.30, stratify=df["category"], random_state=42
)
print(len(train_df), len(test_df))
```
